# PERTANYAAN PRAKTIKUM D3

1. **Apakah perbedaan gambar yang ditampilkan tanpa dan dengan matplotlib?** <br>
↪ Ketika menampilkan gambar menggunakan OpenCV (`cv2_imshow`), gambar ditampilkan dalam format BGR (Blue, Green, Red) secara default. Matplotlib (`plt.imshow`), di sisi lain, menampilkan gambar dalam format RGB (Red, Green, Blue). Ini berarti warna pada gambar akan terlihat berbeda tergantung pada metode yang digunakan, karena urutan channel warnanya terbalik.
2. **Apakah perbedaan dan pengaruhnya pembuatan black image antara tipe data
int16 dan int32?** <br>
↪ Perbedaan utama terletak pada rentang nilai yang dapat disimpan. `int16` dapat menyimpan nilai dari -32768 hingga 32767, sedangkan `int32` dapat menyimpan nilai dari -2147483648 hingga 2147483647. Dalam konteks pembuatan gambar hitam, `int32` menggunakan lebih banyak memori per piksel dibandingkan `int16`.
3. **Apakah kegunaan “google.colab.patches import cv2_imshow” pada potongan
kode berikut** <br>
`from google.colab.patches import cv2_imshow` <br>
`from skimage import io` <br>
↪ `cv2_imshow` adalah fungsi yang disediakan oleh Google Colab untuk menampilkan gambar OpenCV dalam notebook. Fungsi ini diperlukan karena fungsi `cv2.imshow()` bawaan OpenCV tidak dapat berjalan di lingkungan notebook seperti Colab. `cv2_imshow` menangani tampilan gambar dalam output sel notebook.
4. **Apakah kegunaan “skimage import io” pada potongan kode soal nomor 3** <br>
↪ `skimage.io` diimpor untuk penggunaan fungsi-fungsi terkait input/output gambar dari pustaka scikit-image, meskipun dalam sel kode yang ditampilkan saat ini, hanya fungsi dari OpenCV dan Matplotlib yang digunakan untuk membaca dan menampilkan gambar.
